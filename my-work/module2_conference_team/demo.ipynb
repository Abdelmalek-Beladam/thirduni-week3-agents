This notebook displays the saved PowerShell run, state snapshots, and local nested trace. It does not call Gemini or Tavily.

## Architecture

The coordinator stores the brief and delegates evidence review, presentation structure, and rehearsal questions. Each wrapper passes explicit fields from `ConferenceState` to a genuine model-backed specialist. Evidence search is limited to two Tavily searches.

```python
import json
import sys
from pathlib import Path
from IPython.display import Code, Markdown, display

project_root = Path.cwd()
while project_root != project_root.parent and not (project_root / '.venv' / 'Scripts' / 'python.exe').exists():
    project_root = project_root.parent
assert Path(sys.executable).resolve() == (project_root / '.venv' / 'Scripts' / 'python.exe').resolve()
assert sys.version_info[:3] == (3, 13, 1)
project_dir = project_root / 'my-work' / 'module2_conference_team'
output_path = project_dir / 'conference_demo_output.txt'
state_path = project_dir / 'conference_state_snapshots.json'
trace_path = project_dir / 'traces' / 'conference_demo_trace_timeline.txt'
for path in (output_path, state_path, trace_path):
    if not path.exists():
        raise FileNotFoundError(f'Missing saved artifact: {path}')
display(Markdown('## Saved PowerShell output'))
display(Code(output_path.read_text(encoding='utf-8'), language='text'))
display(Markdown('## State snapshots'))
display(Code(state_path.read_text(encoding='utf-8'), language='json'))
display(Markdown('## Nested local callback trace'))
display(Code(trace_path.read_text(encoding='utf-8'), language='text'))
```

## Observations

- Evidence searches used: 2 maximum.
- Outline rebuilt on follow-up: No.
- Questions regenerated for non-technical attendees: Yes.
- Source access/date limitations: Preserved in the evidence specialist output.
- Cloud tracing: Disabled; the displayed trace is local callback telemetry.
