# Week 3 Multi-Agent Systems Demo

This notebook presents the saved PowerShell run and genuine local callback traces. It does not launch MCP, call a model, or contact Tavily.

## Architecture and tracing

Stage 1 uses a supervisor that delegates to model-backed analogy and definition specialists. Stage 3 uses a conference supervisor that delegates evidence requests to a Tavily-backed researcher.

The supervisor message list shows visible delegation calls and tool results. The local callback trace additionally records nested agent and Gemini model runs, including actual run IDs, parent run IDs, and a shared trace ID. This is a local alternative, not the LangSmith UI shown in the course. LangSmith tracing requires an account/API key and `LANGSMITH_TRACING=true`; cloud tracing was not enabled.

```python
import json
import sys
from pathlib import Path
from IPython.display import Code, Markdown, display

project_root = Path.cwd()
while project_root != project_root.parent and not (project_root / '.venv' / 'Scripts' / 'python.exe').exists():
    project_root = project_root.parent
assert Path(sys.executable).resolve() == (project_root / '.venv' / 'Scripts' / 'python.exe').resolve()
assert sys.version_info[:3] == (3, 13, 1)
module_dir = project_root / 'my-work' / 'module2_multi_agent'
output_path = module_dir / 'demo_output.txt'
if not output_path.exists():
    raise FileNotFoundError(f'Run the PowerShell command in README.md first: {output_path}')
display(Markdown('## Saved PowerShell demonstration output'))
display(Code(output_path.read_text(encoding='utf-8'), language='text'))
```

```python
trace_dir = module_dir / 'traces'
for stage in ('stage1', 'stage3'):
    trace_path = trace_dir / f'{stage}_trace.json'
    timeline_path = trace_dir / f'{stage}_trace_timeline.txt'
    if not trace_path.exists() or not timeline_path.exists():
        raise FileNotFoundError(f'Missing saved {stage} trace; run the PowerShell demo first.')
    trace = json.loads(trace_path.read_text(encoding='utf-8'))
    trace_ids = {event['trace_id'] for event in trace['events']}
    assert trace_ids == {trace['trace_id']}
    print(f'{stage}: trace_id={trace["trace_id"]}; events={len(trace["events"])}; shared IDs verified={len(trace_ids) == 1}')
    display(Markdown(f'### {stage} callback timeline'))
    display(Code(timeline_path.read_text(encoding='utf-8'), language='text'))
```

## Observations

- Stage 1 specialist calls: both were observed in the supervisor tool calls.
- Stage 3 delegation: Tavily returned several sources. The final answer's Applied AI Course and Medium URLs both matched returned results. The Applied AI page was accessible and its excerpt partially supported generalization via cross-validation, but no publication date was supplied and it is an educational source, not a peer-reviewed paper. The Medium result returned HTTP 403 and remains unverified.
- Evidence limitations: The final answer did not preserve each source's different status/limitation, although it said the sources had not been independently verified. It did not claim the user's model passed external validation and did not invent a numeric result.
- Detector note: The first run printed that a numeric claim was present because the broad check matched “accuracy” in the disclaimer. The checker was narrowed afterward to numeric metrics/percentages; that revision was not rerun to avoid another model call.
- Limitation: The local callback trace is actual execution telemetry, but it is not the course's cloud LangSmith UI.
